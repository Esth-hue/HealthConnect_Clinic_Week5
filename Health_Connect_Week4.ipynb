### Validate Previous No-Show History
A patient's number of previous no-shows should not exceed their number of previous appointments. This counts records that violate that logical relationship.

### Validate Appointment and Booking Dates
An appointment should normally occur after it is booked. This counts records where the appointment date is later than the booking date.

### Validate the Booking Lead-Time Field
The actual lead time is independently calculated by subtracting the booking date from the appointment date. It is then compared with the existing "booking_lead_days" column to confirm that the recorded values are accurate.

```python
calculated_lead_days = (df["appointment_date"] - df["booking_date"]).dt.days
(df["booking_lead_days"] != calculated_lead_days).sum()
```

This means there were zero discrepancies across all 5,000 records. Therefore, the booking_lead_days variable is completely consistent with the underlying booking and appointment dates.

### Check for Impossible Date Order

### Examine Reminder-Channel Missingness

### Count Appointment Outcomes

### Analyse Outcomes by Appointment Type

### Calculate Outcome Rates by Appointment Type
This cross-tabulation calculates the percentage of each outcome within every appointment type. It allows fair comparison between appointment types even when their total volumes differ.

```python
pd.crosstab(
    df["appointment_type"],
    df["appointment_outcome"],
    normalize="index"
) * 100
```

The highest no-show rate was observed among Follow-up appointments at 51.23%. Diagnostic Tests had the second-highest no-show rate at 49.75%, followed by Specialist Consultations at 47.44%. General Consultations had the lowest no-show rate at 46.64%.
This indicates that appointment type may be associated with differences in attendance behaviour, although the differences are not extremely large.
An important observation is that Follow-up appointments have a particularly high no-show rate despite General Consultations having the largest number of appointments. This means that the result is not simply being driven by appointment volume.

### Calculate Outcome Rates by Reminder Status

These results indicate an observed association between reminder status and appointment outcomes. However, this analysis does not establish causation, and other appointment or patient characteristics may also influence the relationship.

### Create Booking Lead-Time Groups
Continuous booking lead days are converted into meaningful categories using. The groups make it easier to compare appointment outcomes across practical booking windows: 0–7, 8–14, 15–30, 31–45, and 46–60 days.

```python
df["lead_time_group"] = pd.cut(
    df["booking_lead_days"],
    bins=[-1, 7, 14, 30, 45, 60],
    labels=["0-7 days", "8-14 days", "15-30 days", "31-45 days", "46-60 days"]
)
```

### Count Appointments by Lead-Time Group
This counts how many appointments fall within each booking lead-time category. Sorting by category order keeps the groups in their natural progression.

```python
df["lead_time_group"].value_counts().sort_index()
```

### Calculate Outcomes by Lead-Time Group
The cross-tabulation calculates the percentage of each appointment outcome within every lead-time group. This helps identify whether booking further in advance is associated with higher or lower no-show rates.

```python
pd.crosstab(
    df["lead_time_group"],
    df["appointment_outcome"],
    normalize="index"
) * 100
```

### Compare Raw Outcomes Across Lead-TimeGgroups
This cross-tabulation provides the actual number of attended, no-show, and cancelled appointments in each lead-time category. The raw counts complement the percentage analysis above.

```python
pd.crosstab(
    df["lead_time_group"],
    df["appointment_outcome"]
)
```

### Validate the New Lead-Time Grouping
This checks whether any records failed to receive a lead-time category. A zero result confirms that all booking lead-day values were successfully assigned to one of the defined groups.

```python
df["lead_time_group"].isnull().sum()
```

Booking lead time was further analysed by grouping appointments into five intervals between 0 and 60 days. A clear and consistent increase in no-show rates was observed as booking lead time increased. Appointments booked 0–7 days in advance had a no-show rate of 27.81%, increasing to 33.55% for 8–14 days, 43.21% for 15–30 days, 53.82% for 31–45 days, and 67.69% for appointments booked 46–60 days in advance. This represents a 39.88-percentage-point difference between the shortest and longest lead-time groups. The consistent increase across every interval indicates that booking lead time is one of the strongest observed factors associated with appointment no-shows in the dataset. However, this remains an observational relationship and should not be interpreted as proof that longer booking lead times directly cause no-shows.
